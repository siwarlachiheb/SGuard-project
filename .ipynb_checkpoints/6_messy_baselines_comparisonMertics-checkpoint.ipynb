# OMIn GraphRAG — Notebook 6: Baselines and Comparison

**Systems evaluated:**
1. Vanilla LLM — direct generation, no retrieval
2. Naive RAG — TF-IDF text chunk retrieval
3. KEO-style KG-RAG — flat weights, no guardrails
4. Semantic KG-RAG — cosine weighting, no guardrails (notebook 4 baseline)
5. Post-Guardrail KG-RAG — semantic + post-generation only
6. Full System — semantic + both guardrails (notebook 5)
7. Self-RAG style — LLM self-critique baseline

**Metrics (all deterministic except LLM-judge):**
- Domain classification accuracy
- Abstention accuracy (unanswerable + OOD)
- False abstention rate (answerable)
- Grounding score (post-guardrail)
- Hallucination rate
- Pre-guardrail schema match score
- Interception rate
- ROUGE-1 and ROUGE-L F1
- Retrieval latency
- Entity match rate

## 2. Shared utility functions

````python
def safe_parse_json(raw):
    raw = raw.strip().replace("```json","").replace("```","").strip()
    for strategy in [
        lambda x: json.loads(x),
        lambda x: json.loads(re.search(r'\{.*\}', x, re.DOTALL).group()),
        lambda x: json.loads(re.sub(r',(\s*[}\]])', r'\1', x))
    ]:
        try:
            return strategy(raw)
        except:
            pass
    return None

def compute_rouge(answer, reference):
    if not answer or not reference:
        return {"rouge1": 0.0, "rougeL": 0.0}
    scores = scorer.score(reference, answer)
    return {
        "rouge1": scores["rouge1"].fmeasure,
        "rougeL": scores["rougeL"].fmeasure
    }

def compute_entity_match(answer, expected_contains):
    """Fraction of expected entities found in answer (case-insensitive)."""
    if not expected_contains:
        return None
    answer_lower = answer.lower()
    found = sum(1 for e in expected_contains if e.lower() in answer_lower)
    return found / len(expected_contains)

def gemini_call_with_retry(client, prompt, model=GENERATION_MODEL, temp=0.2, max_retries=3):
    for attempt in range(max_retries):
        try:
            r = client.models.generate_content(
                model=model, contents=prompt,
                config={"temperature": temp}
            )
            return r.text.strip()
        except Exception as e:
            if "503" in str(e):
                wait = 30 * (attempt + 1)
                print(f"  503, waiting {wait}s")
                time.sleep(wait)
            else:
                return None
    return None

def embed_query(query, client, model=EMBEDDING_MODEL):
    result = client.models.embed_content(model=model, contents=[query])
    emb = np.array(result.embeddings[0].values, dtype=np.float32)
    return emb / (np.linalg.norm(emb) + 1e-9)

def check_domain(query, client):
    prompt = """Classify as IN_DOMAIN (aviation maintenance) or OUT_OF_DOMAIN.
Query: "{query}"
JSON only: {{"domain": "IN_DOMAIN" or "OUT_OF_DOMAIN", "answerable": "LIKELY" or "UNLIKELY", "reason": "one sentence"}}""".format(query=query)
    raw = gemini_call_with_retry(client, prompt, temp=0.0)
    result = safe_parse_json(raw) if raw else None
    return result or {"domain": "IN_DOMAIN", "answerable": "LIKELY", "reason": "filter failed"}

print("Utility functions ready.")
````

## 3. Baseline 1: Vanilla LLM

````python
VANILLA_PROMPT = """You are an aviation maintenance expert.
Answer the following question about aviation maintenance based on your knowledge.
If you don't know, say so explicitly.

Question: {query}
Answer:"""

def vanilla_llm_query(query, client):
    start   = time.time()
    domain  = check_domain(query, client)
    if domain["domain"] == "OUT_OF_DOMAIN":
        return {"answer": f"Out of domain: {domain['reason']}",
                "abstained": True, "abstain_reason": "out_of_domain",
                "domain_detected": "OUT_OF_DOMAIN", "latency": time.time()-start,
                "records_retrieved": 0}
    prompt = VANILLA_PROMPT.format(query=query)
    answer = gemini_call_with_retry(client, prompt) or "Error generating answer."
    # Simple self-check: did LLM say it doesn't know?
    abstained = any(phrase in answer.lower() for phrase in
                    ["i don't know", "i cannot", "not enough information",
                     "i'm not sure", "cannot provide"])
    return {"answer": answer, "abstained": abstained,
            "abstain_reason": "self_reported" if abstained else None,
            "domain_detected": "IN_DOMAIN", "latency": time.time()-start,
            "records_retrieved": 0}

# Quick test
test = vanilla_llm_query("What causes hydraulic leaks in landing gear?", client)
print(f"Vanilla LLM test:")
print(f"  Answer: {test['answer'][:200]}")
print(f"  Abstained: {test['abstained']}")
print(f"  Latency: {test['latency']:.2f}s")
````

## 4. Baseline 2: Naive RAG (TF-IDF text chunk retrieval)

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Build TF-IDF index over all records
print("Building TF-IDF index...")
record_texts = [r["text_original"] for r in records]
record_ids   = [r["record_id"] for r in records]

tfidf = TfidfVectorizer(max_features=10000, stop_words="english",
                         ngram_range=(1,2))
tfidf_matrix = tfidf.fit_transform(record_texts)
print(f"TF-IDF index built: {tfidf_matrix.shape}")

NAIVE_RAG_PROMPT = """You are an aviation maintenance expert assistant.
Answer the question based ONLY on the provided records.
If records don't contain sufficient information, say so.

Question: {query}

Retrieved Records:
{context}

Answer:"""

def naive_rag_query(query, client, tfidf, tfidf_matrix, records, top_k=15):
    start  = time.time()
    domain = check_domain(query, client)
    if domain["domain"] == "OUT_OF_DOMAIN":
        return {"answer": f"Out of domain: {domain['reason']}",
                "abstained": True, "domain_detected": "OUT_OF_DOMAIN",
                "latency": time.time()-start, "records_retrieved": 0}

    # TF-IDF retrieval
    query_vec = tfidf.transform([query])
    sims      = cosine_similarity(query_vec, tfidf_matrix).flatten()
    top_idx   = np.argsort(sims)[::-1][:top_k]
    retrieved = []
    for idx in top_idx:
        if sims[idx] > 0:
            rec = records[idx]
            retrieved.append(f"[{rec['record_id']} | {rec['source_file']}]\n{rec['text_original']}")

    if not retrieved:
        return {"answer": "No relevant records found.", "abstained": True,
                "domain_detected": "IN_DOMAIN", "latency": time.time()-start,
                "records_retrieved": 0}

    context = "\n\n".join(retrieved)
    prompt  = NAIVE_RAG_PROMPT.format(query=query, context=context[:4000])
    answer  = gemini_call_with_retry(client, prompt) or "Error generating answer."
    abstained = any(phrase in answer.lower() for phrase in
                    ["do not contain", "not enough information", "cannot answer",
                     "insufficient", "no information"])
    return {"answer": answer, "abstained": abstained,
            "domain_detected": "IN_DOMAIN", "latency": time.time()-start,
            "records_retrieved": len(retrieved)}

# Test
test = naive_rag_query("What causes hydraulic leaks in landing gear?",
                        client, tfidf, tfidf_matrix, records)
print(f"Naive RAG test:")
print(f"  Answer: {test['answer'][:200]}")
print(f"  Records: {test['records_retrieved']}, Latency: {test['latency']:.2f}s")
````

## 5. Baseline 3: KEO-style KG-RAG (flat weights, no guardrails)

````python
def find_seed_nodes_flat(query_emb, entity_matrix_norm, entity_list, G, top_k=10):
    """Simple top-k by cosine similarity — no MMR diversity."""
    sims = entity_matrix_norm @ query_emb
    top_idx = np.argsort(sims)[::-1]
    seeds = []
    for idx in top_idx:
        entity = entity_list[idx]
        if entity in G.nodes:
            seeds.append({"node": entity, "similarity": float(sims[idx])})
        if len(seeds) >= top_k:
            break
    return seeds

def expand_subgraph_flat(seed_nodes, G, hops=2):
    """Same expansion but returns ALL edges with flat weight=1."""
    visited  = set(s["node"] for s in seed_nodes)
    frontier = set(visited)
    for _ in range(hops):
        new_frontier = set()
        for node in frontier:
            if node not in G: continue
            for nb in list(G.successors(node)) + list(G.predecessors(node)):
                if nb not in visited:
                    new_frontier.add(nb)
                    visited.add(nb)
        frontier = new_frontier
    edges = []
    for u, v, data in G.edges(data=True):
        if u in visited and v in visited:
            edges.append({"head": u, "tail": v, "relation": data["relation"],
                          "record_ids": data["record_ids"],
                          "support_count": data["support_count"],
                          "relevance_score": 1.0})  # flat weight
    return edges, visited

KEO_PROMPT = """You are an aviation maintenance expert.
Answer the question using the knowledge graph context below.
If insufficient, say so.

Question: {query}

Knowledge Graph Context:
{kg_context}

Answer:"""

def keo_style_query(query, client, G, entity_matrix_norm, entity_list,
                     entity_embeddings, record_lookup, top_k=10, hops=2):
    start  = time.time()
    domain = check_domain(query, client)
    if domain["domain"] == "OUT_OF_DOMAIN":
        return {"answer": f"Out of domain: {domain['reason']}",
                "abstained": True, "domain_detected": "OUT_OF_DOMAIN",
                "latency": time.time()-start, "records_retrieved": 0,
                "edges_in_subgraph": 0}

    query_emb = embed_query(query, client)
    seeds     = find_seed_nodes_flat(query_emb, entity_matrix_norm, entity_list, G, top_k)
    edges, _  = expand_subgraph_flat(seeds, G, hops)

    # KEO-style: use ALL edges as context (no semantic filtering)
    # Sort by support_count (frequency) as KEO does
    edges_sorted = sorted(edges, key=lambda x: -x["support_count"])[:50]

    # Convert to text (KEO-style context reconstruction)
    kg_lines = [f"{e['head']} --[{e['relation']}]--> {e['tail']}" for e in edges_sorted]
    kg_context = "\n".join(kg_lines)

    # Also retrieve records
    record_scores = defaultdict(float)
    for e in edges_sorted:
        for rid in e["record_ids"]:
            record_scores[rid] += e["support_count"]
    top_recs = sorted(record_scores.items(), key=lambda x: -x[1])[:15]
    rec_context = ""
    for rid, _ in top_recs:
        if rid in record_lookup:
            rec_context += f"[{rid}] {record_lookup[rid]['text_original']}\n"

    full_context = f"GRAPH EDGES:\n{kg_context}\n\nSOURCE RECORDS:\n{rec_context}"
    prompt = KEO_PROMPT.format(query=query, kg_context=full_context[:4000])
    answer = gemini_call_with_retry(client, prompt) or "Error."
    abstained = any(p in answer.lower() for p in
                    ["not enough", "cannot answer", "insufficient", "do not contain"])
    return {"answer": answer, "abstained": abstained,
            "domain_detected": "IN_DOMAIN", "latency": time.time()-start,
            "records_retrieved": len(top_recs), "edges_in_subgraph": len(edges)}

# Test
test = keo_style_query("What causes hydraulic leaks in landing gear?",
                        client, G, entity_matrix_norm, entity_list,
                        entity_embeddings, record_lookup)
print(f"KEO-style test:")
print(f"  Edges in subgraph: {test['edges_in_subgraph']}")
print(f"  Answer: {test['answer'][:200]}")
print(f"  Latency: {test['latency']:.2f}s")
````

## 6. Baseline 4: Self-RAG style (self-critique)

````python
SELFRAG_RETRIEVE_PROMPT = """You are an aviation maintenance expert.
First decide: do you need to retrieve information to answer this query?
Query: "{query}"
JSON only: {{"need_retrieval": true or false, "reason": "why"}}"""

SELFRAG_CRITIQUE_PROMPT = """Evaluate this answer to the aviation maintenance query.
Query: {query}
Retrieved context: {context}
Answer: {answer}

Rate on:
- isrel: is the retrieved context relevant? (true/false)
- issup: is the answer supported by the context? (true/false)
- isuse: is the answer useful? (1-5)

JSON only: {{"isrel": bool, "issup": bool, "isuse": 1-5}}"""

def selfrag_style_query(query, client, tfidf, tfidf_matrix, records, top_k=10):
    """Self-RAG inspired: decide retrieval need, generate, self-critique."""
    start  = time.time()
    domain = check_domain(query, client)
    if domain["domain"] == "OUT_OF_DOMAIN":
        return {"answer": f"Out of domain.", "abstained": True,
                "domain_detected": "OUT_OF_DOMAIN", "latency": time.time()-start,
                "records_retrieved": 0, "critique": None}

    # Step 1: decide if retrieval needed
    raw = gemini_call_with_retry(client, SELFRAG_RETRIEVE_PROMPT.format(query=query), temp=0.0)
    retrieve_decision = safe_parse_json(raw) if raw else {"need_retrieval": True}

    context = ""
    n_retrieved = 0
    if retrieve_decision.get("need_retrieval", True):
        query_vec = tfidf.transform([query])
        sims      = cosine_similarity(query_vec, tfidf_matrix).flatten()
        top_idx   = np.argsort(sims)[::-1][:top_k]
        ctx_parts = []
        for idx in top_idx:
            if sims[idx] > 0:
                rec = records[idx]
                ctx_parts.append(f"[{rec['record_id']}] {rec['text_original']}")
        context     = "\n".join(ctx_parts)
        n_retrieved = len(ctx_parts)

    # Step 2: generate answer
    gen_prompt = NAIVE_RAG_PROMPT.format(query=query, context=context[:3000]) if context \
                 else VANILLA_PROMPT.format(query=query)
    answer = gemini_call_with_retry(client, gen_prompt) or "Error."

    # Step 3: self-critique
    crit_prompt = SELFRAG_CRITIQUE_PROMPT.format(
        query=query, context=context[:1000], answer=answer[:500])
    raw_crit  = gemini_call_with_retry(client, crit_prompt, temp=0.0)
    critique  = safe_parse_json(raw_crit) if raw_crit else None

    # Abstain if critique says not supported
    abstained = False
    if critique and not critique.get("issup", True):
        answer    = f"[Self-critique: answer not supported by evidence] {answer}"
        abstained = True

    return {"answer": answer, "abstained": abstained,
            "domain_detected": "IN_DOMAIN", "latency": time.time()-start,
            "records_retrieved": n_retrieved, "critique": critique}

# Test
test = selfrag_style_query("What causes hydraulic leaks in landing gear?",
                            client, tfidf, tfidf_matrix, records)
print(f"Self-RAG style test:")
print(f"  Answer: {test['answer'][:200]}")
print(f"  Critique: {test['critique']}")
print(f"  Latency: {test['latency']:.2f}s")
````

## BYOKGRAG

````python
# ============================================================
# BYOKG-RAG Simplified Baseline
# Based on: Mavromatis et al. (2025) "BYOKG-RAG: Multi-Strategy
# Graph Retrieval for Knowledge Graph Question Answering"
# Amazon Science. https://cdn.amazon.science/7f/72/...
#
# Implemented components:
#   - KG-Linker: entity extraction + path identification + draft answer
#   - Multi-strategy entity linking: embedding similarity + string match
#   - Path retrieval: BFS from linked entities + shortest path scoring
#   - Scoring-based triplet retrieval: cosine similarity ranking
#
# Omitted (require graph DB infrastructure):
#   - OpenCypher query generation (requires Neo4j/Neptune)
#   - Iterative refinement loop (TR=2 iterations)
#   - Agentic step-by-step traversal
# ============================================================

import difflib

BYOKG_LINKER_PROMPT = """You are a knowledge graph expert for aviation maintenance.
Given a question and the graph schema, generate graph artifacts for retrieval.

GRAPH SCHEMA:
- Node types: aviation entities (components, failures, actions, conditions)
- Relation types: CAUSES, RESULTS_IN, PART_OF, LOCATED_IN, REQUIRES_ACTION,
  INSPECTED_BY, REPLACED_BY, PRECEDED_BY, AFFECTS, INSTANCE_OF

GRAPH CONTEXT (if available):
{graph_context}

Question: {question}

Perform these tasks:
1. ENTITY EXTRACTION: Extract all topic entities from the question
2. PATH IDENTIFICATION: Identify relation paths that connect entities to answers
3. DRAFT ANSWER: Generate a draft answer using your knowledge

Respond ONLY with JSON:
{{
  "entities": ["list of extracted entities from question"],
  "relation_paths": [["RELATION1", "RELATION2"], ["RELATION1"]],
  "draft_answer_entities": ["list of likely answer entities"],
  "draft_answer": "brief draft answer"
}}"""


def byokg_entity_linking(generated_entities, G, entity_embeddings,
                          entity_matrix_norm, entity_list, entity_to_idx,
                          top_m=3):
    """
    Multi-strategy entity linking: embedding similarity + fuzzy string match.
    Maps LLM-generated entity mentions to actual graph nodes.
    """
    linked = set()

    for gen_entity in generated_entities:
        gen_lower = gen_entity.lower().strip()

        # Strategy 1: Embedding similarity
        if gen_entity in entity_embeddings:
            emb = np.array(entity_embeddings[gen_entity], dtype=np.float32)
            emb = emb / (np.linalg.norm(emb) + 1e-9)
            sims = entity_matrix_norm @ emb
            top_idx = np.argsort(sims)[::-1][:top_m]
            for idx in top_idx:
                node = entity_list[idx]
                if node in G.nodes and sims[idx] > 0.7:
                    linked.add(node)
        else:
            # Embed the generated entity on the fly
            try:
                result = client.models.embed_content(
                    model=EMBEDDING_MODEL, contents=[gen_entity])
                emb = np.array(result.embeddings[0].values, dtype=np.float32)
                emb = emb / (np.linalg.norm(emb) + 1e-9)
                sims = entity_matrix_norm @ emb
                top_idx = np.argsort(sims)[::-1][:top_m]
                for idx in top_idx:
                    node = entity_list[idx]
                    if node in G.nodes and sims[idx] > 0.7:
                        linked.add(node)
            except:
                pass

        # Strategy 2: Fuzzy string matching
        graph_nodes = list(G.nodes())
        matches = difflib.get_close_matches(gen_lower, graph_nodes,
                                             n=top_m, cutoff=0.6)
        for m in matches:
            linked.add(m)

    return list(linked)


def byokg_path_retrieval(linked_entities, draft_answer_entities, G,
                          max_paths=20):
    """
    Path retrieval: BFS expansion + shortest paths between
    linked entities and draft answer entities.
    Implements Follow-Paths and Shortest-Paths from BYOKG-RAG.
    """
    retrieved_triplets = []
    visited_edges = set()

    # BFS from linked entities (1-hop)
    for entity in linked_entities[:5]:  # cap to 5 seed entities
        if entity not in G:
            continue
        for u, v, data in G.edges(entity, data=True):
            key = (u, data["relation"], v)
            if key not in visited_edges:
                visited_edges.add(key)
                retrieved_triplets.append({
                    "head": u, "relation": data["relation"], "tail": v,
                    "record_ids": data["record_ids"],
                    "support_count": data["support_count"],
                    "retrieval_method": "bfs_expansion"
                })
        for u, v, data in G.in_edges(entity, data=True):
            key = (u, data["relation"], v)
            if key not in visited_edges:
                visited_edges.add(key)
                retrieved_triplets.append({
                    "head": u, "relation": data["relation"], "tail": v,
                    "record_ids": data["record_ids"],
                    "support_count": data["support_count"],
                    "retrieval_method": "bfs_expansion"
                })

    # Shortest paths between linked entities and draft answer entities
    all_source = linked_entities[:3]
    all_target = [e for e in draft_answer_entities if e in G][:3]

    undirected = G.to_undirected()
    for src in all_source:
        for tgt in all_target:
            if src == tgt or src not in G or tgt not in G:
                continue
            try:
                path_nodes = nx.shortest_path(undirected, src, tgt)
                for i in range(len(path_nodes) - 1):
                    u, v = path_nodes[i], path_nodes[i+1]
                    for _, _, data in G.edges(u, data=True):
                        if data.get("tail", v) == v or True:
                            key = (u, data["relation"], v)
                            if key not in visited_edges:
                                visited_edges.add(key)
                                retrieved_triplets.append({
                                    "head": u, "relation": data["relation"],
                                    "tail": v, "record_ids": data["record_ids"],
                                    "support_count": data["support_count"],
                                    "retrieval_method": "shortest_path"
                                })
            except (nx.NetworkXNoPath, nx.NodeNotFound):
                pass

    return retrieved_triplets[:max_paths]


def byokg_scoring_retrieval(query_emb, retrieved_triplets,
                              entity_embeddings, relation_embeddings, top_k=30):
    """
    Scoring-based triplet retrieval: rank by semantic similarity to query.
    Implements the scoring-based retrieval from BYOKG-RAG Section 3.5.
    """
    scored = []
    for t in retrieved_triplets:
        head_emb = entity_embeddings.get(t["head"])
        tail_emb = entity_embeddings.get(t["tail"])
        rel_emb  = relation_embeddings.get(t["relation"])
        if head_emb is None or tail_emb is None or rel_emb is None:
            continue
        h = np.array(head_emb, dtype=np.float32)
        t_e = np.array(tail_emb, dtype=np.float32)
        r = np.array(rel_emb, dtype=np.float32)
        h /= (np.linalg.norm(h) + 1e-9)
        t_e /= (np.linalg.norm(t_e) + 1e-9)
        r /= (np.linalg.norm(r) + 1e-9)
        score = (0.4 * float(np.dot(query_emb, h)) +
                 0.4 * float(np.dot(query_emb, t_e)) +
                 0.2 * float(np.dot(query_emb, r)))
        scored.append({**t, "relevance_score": score})

    scored.sort(key=lambda x: -x["relevance_score"])
    return scored[:top_k]


BYOKG_GENERATION_PROMPT = """You are an aviation maintenance expert.
Answer the question using the retrieved knowledge graph context and source records.
If insufficient, say so explicitly.

Question: {query}

KG-Linker Draft Answer: {draft_answer}

Retrieved Graph Triplets:
{triplet_context}

Source Records:
{record_context}

Final Answer:"""


def byokg_rag_query(query, client, G, entity_matrix_norm, entity_list,
                     entity_to_idx, entity_embeddings, relation_embeddings,
                     record_lookup, graph_schema):
    """
    BYOKG-RAG simplified pipeline.
    Returns result dict compatible with baseline evaluation.
    """
    start  = time.time()
    domain = check_domain(query, client)

    if domain["domain"] == "OUT_OF_DOMAIN":
        return {"answer": f"Out of domain: {domain['reason']}",
                "abstained": True, "domain_detected": "OUT_OF_DOMAIN",
                "latency": time.time()-start, "records_retrieved": 0,
                "linked_entities": [], "retrieval_methods": []}

    # Step 1: KG-Linker — generate graph artifacts
    linker_prompt = BYOKG_LINKER_PROMPT.format(
        question=query, graph_context=""
    )
    raw = gemini_call_with_retry(client, linker_prompt, temp=0.1)
    artifacts = safe_parse_json(raw) if raw else {}

    generated_entities      = artifacts.get("entities", [])
    draft_answer_entities   = artifacts.get("draft_answer_entities", [])
    draft_answer            = artifacts.get("draft_answer", "")
    relation_paths          = artifacts.get("relation_paths", [])

    # Step 2: Multi-strategy entity linking
    linked_entities = byokg_entity_linking(
        generated_entities + draft_answer_entities,
        G, entity_embeddings, entity_matrix_norm,
        entity_list, entity_to_idx, top_m=3
    )

    # Fallback: embed query if no entities linked
    if not linked_entities:
        query_emb_fallback = embed_query(query, client)
        sims = entity_matrix_norm @ query_emb_fallback
        top_idx = np.argsort(sims)[::-1][:5]
        linked_entities = [entity_list[i] for i in top_idx if entity_list[i] in G.nodes]

    # Step 3: Path retrieval (BFS + shortest paths)
    path_triplets = byokg_path_retrieval(
        linked_entities, draft_answer_entities, G, max_paths=50
    )

    # Step 4: Scoring-based retrieval
    query_emb = embed_query(query, client)
    scored_triplets = byokg_scoring_retrieval(
        query_emb, path_triplets, entity_embeddings, relation_embeddings, top_k=30
    )

    # Step 5: Retrieve source records
    record_scores = defaultdict(float)
    for t in scored_triplets:
        for rid in t.get("record_ids", []):
            record_scores[rid] += t["relevance_score"]

    top_recs = sorted(record_scores.items(), key=lambda x: -x[1])[:15]
    retrieved_records = []
    for rid, score in top_recs:
        if rid in record_lookup:
            rec = record_lookup[rid]
            retrieved_records.append(
                f"[{rid} | {rec['source_file']}] {rec['text_original']}"
            )

    # Step 6: Generate final answer
    triplet_context = "\n".join(
        f"{t['head']} --[{t['relation']}]--> {t['tail']} "
        f"(score={t['relevance_score']:.3f}, via {t['retrieval_method']})"
        for t in scored_triplets[:15]
    )
    record_context = "\n".join(retrieved_records[:10])

    gen_prompt = BYOKG_GENERATION_PROMPT.format(
        query=query, draft_answer=draft_answer,
        triplet_context=triplet_context, record_context=record_context[:3000]
    )
    answer = gemini_call_with_retry(client, gen_prompt) or "Error generating answer."
    abstained = any(p in answer.lower() for p in
                    ["not enough", "cannot answer", "insufficient",
                     "do not contain", "no information"])

    return {
        "answer": answer,
        "abstained": abstained,
        "domain_detected": "IN_DOMAIN",
        "latency": time.time() - start,
        "records_retrieved": len(top_recs),
        "linked_entities": linked_entities[:5],
        "draft_answer": draft_answer,
        "retrieval_methods": list(set(t["retrieval_method"] for t in scored_triplets))
    }


# Test BYOKG-RAG
print("=== BYOKG-RAG SIMPLIFIED TEST ===")
test = byokg_rag_query(
    "What causes hydraulic leaks in landing gear?",
    client, G, entity_matrix_norm, entity_list, entity_to_idx,
    entity_embeddings, relation_embeddings, record_lookup, graph_schema
)
print(f"Linked entities: {test['linked_entities']}")
print(f"Draft answer:    {test['draft_answer'][:150]}")
print(f"Retrieval methods: {test['retrieval_methods']}")
print(f"Records retrieved: {test['records_retrieved']}")
print(f"Answer: {test['answer'][:300]}")
print(f"Latency: {test['latency']:.2f}s")
````

## 8. Compute full comparison table with deterministic metrics

````python
# ============================================================
# OPTION 2: LLM-as-Judge scoring on answerable questions
# Judge: Gemini 2.5 Flash evaluating answer quality 1-5
# Criteria adapted from KEO paper evaluation metrics
# ============================================================

LLM_JUDGE_PROMPT = """You are an expert aviation maintenance evaluator.
Score the following answer to an aviation maintenance question on a 1-5 scale.

QUESTION: {question}

ANSWER TO EVALUATE: {answer}

REFERENCE CONTEXT (what a good answer should cover):
{reference}

Score on these criteria (1-5 each):
1. CORRECTNESS: Is the answer factually accurate for aviation maintenance?
2. COMPLETENESS: Does it cover the key aspects needed to answer the question?
3. GROUNDING: Is the answer specific and grounded, or vague and generic?
4. SAFETY: Would following this answer maintain aviation safety standards?
5. CLARITY: Is the answer clear and actionable?

Respond ONLY with JSON:
{{
  "correctness": 1-5,
  "completeness": 1-5,
  "grounding": 1-5,
  "safety": 1-5,
  "clarity": 1-5,
  "overall": 1-5,
  "reasoning": "one sentence explanation"
}}"""

def llm_judge_answer(question, answer, reference_answers, client):
    """Score an answer using Gemini as judge."""
    reference = "\n".join(reference_answers[:3]) if reference_answers else "No reference available."
    prompt = LLM_JUDGE_PROMPT.format(
        question=question,
        answer=answer[:800],
        reference=reference[:500]
    )
    for attempt in range(3):
        try:
            r = client.models.generate_content(
                model=GENERATION_MODEL, contents=prompt,
                config={"temperature": 0.0}
            )
            result = safe_parse_json(r.text)
            if result:
                return result
        except Exception as e:
            if "503" in str(e):
                time.sleep(30 * (attempt + 1))
            else:
                break
    return {"correctness": 3, "completeness": 3, "grounding": 3,
            "safety": 3, "clarity": 3, "overall": 3,
            "reasoning": "judge failed"}

# Build reference answers from full system's answers
# (used as context for the judge, not as gold standard)
full_sys_answers = {}
for qtype in ["in_domain_answerable", "multi_hop"]:
    for r in guarded_results.get(qtype, []):
        if not r.get("abstained"):
            full_sys_answers[r["question"]] = r["answer"]

JUDGE_CHECKPOINT = "./output/llm_judge_results.json"

if os.path.exists(JUDGE_CHECKPOINT):
    with open(JUDGE_CHECKPOINT) as f:
        judge_results = json.load(f)
    print(f"Resumed judge checkpoint: {sum(len(v) for v in judge_results.values())} done")
else:
    judge_results = {}

# Evaluate all systems on answerable + multi-hop questions
systems_to_judge = {
    "vanilla_llm":    all_baseline_results.get("vanilla_llm", {}),
    "naive_rag":      all_baseline_results.get("naive_rag", {}),
    "keo_style":      all_baseline_results.get("keo_style", {}),
    "selfrag_style":  all_baseline_results.get("selfrag_style", {}),
    "byokg_rag":      all_baseline_results.get("byokg_rag", {}),
    "semantic_kgrag": semantic_baseline_results,
    "full_system":    guarded_results,
}

JUDGE_QTYPES = ["in_domain_answerable", "multi_hop"]

for sys_name, results_by_type in systems_to_judge.items():
    if sys_name not in judge_results:
        judge_results[sys_name] = {}

    print(f"\n=== Judging: {sys_name} ===")

    for qtype in JUDGE_QTYPES:
        results = results_by_type.get(qtype, [])
        if not results:
            continue

        already_done = set(judge_results[sys_name].get(qtype, {}).keys())
        if qtype not in judge_results[sys_name]:
            judge_results[sys_name][qtype] = {}

        to_judge = [r for r in results
                    if r.get("question") not in already_done
                    and not r.get("abstained", False)]

        if not to_judge:
            print(f"  {qtype}: already complete or all abstained")
            continue

        print(f"  {qtype}: judging {len(to_judge)} answers")

        for i, r in enumerate(tqdm(to_judge, desc=f"{sys_name}/{qtype}")):
            question = r.get("question", "")
            answer   = r.get("answer", "")
            ref      = [full_sys_answers.get(question, "")]

            scores = llm_judge_answer(question, answer, ref, client)
            judge_results[sys_name][qtype][question] = {
                "scores": scores,
                "answer_preview": answer[:200]
            }
            time.sleep(2.0)

            if (i + 1) % 5 == 0:
                with open(JUDGE_CHECKPOINT, "w") as f:
                    json.dump(judge_results, f, indent=2)
                tqdm.write(f"  Judge checkpoint saved")

with open(JUDGE_CHECKPOINT, "w") as f:
    json.dump(judge_results, f, indent=2)
print("\nJudge results saved.")

# Print summary table
print("\n=== LLM-AS-JUDGE RESULTS (1-5 scale) ===")
print(f"{'System':<20} {'Correct':>9} {'Complete':>9} {'Grounding':>10} {'Safety':>8} {'Clarity':>8} {'Overall':>8}")
print("-" * 80)

for sys_name in systems_to_judge:
    all_scores = {"correctness": [], "completeness": [], "grounding": [],
                  "safety": [], "clarity": [], "overall": []}

    for qtype in JUDGE_QTYPES:
        for q, data in judge_results.get(sys_name, {}).get(qtype, {}).items():
            scores = data.get("scores", {})
            for metric in all_scores:
                val = scores.get(metric)
                if val is not None:
                    try:
                        all_scores[metric].append(float(val))
                    except:
                        pass

    if not all_scores["overall"]:
        print(f"{sys_name:<20} {'N/A':>9}")
        continue

    row = f"{sys_name:<20}"
    for metric in ["correctness", "completeness", "grounding", "safety", "clarity", "overall"]:
        vals = all_scores[metric]
        row += f" {np.mean(vals):>8.2f}" if vals else f" {'N/A':>8}"
    print(row)

with open("./output/judge_summary.json", "w") as f:
    json.dump({sys: {qt: {q: d["scores"] for q, d in qtdata.items()}
                     for qt, qtdata in qtypes.items()}
               for sys, qtypes in judge_results.items()}, f, indent=2)
print("\nSaved to ./output/judge_summary.json")
````

````python
IDEAL_SCHEMA_PROMPT = """You are an aviation maintenance knowledge graph expert.
Given a query, generate the IDEAL SUBSCHEMA — the minimum set of entity types
and relation types that a knowledge graph MUST contain to answer this query correctly.

Available relation types:
CAUSES, RESULTS_IN, PART_OF, LOCATED_IN, REQUIRES_ACTION,
INSPECTED_BY, REPLACED_BY, PRECEDED_BY, AFFECTS, INSTANCE_OF

Graph hub entities for context (most connected nodes):
{hub_entities}

Query: "{query}"

Generate the ideal subschema as JSON:
{{
  "canonical_form": "structured restatement of query intent in 1 sentence",
  "required_entity_types": ["list of entity concept types needed, e.g. 'hydraulic component', 'failure cause'"],
  "required_relations": ["list of relation types from the available set that must be present"],
  "ideal_triplet_patterns": [
    {{"head_type": "type", "relation": "RELATION", "tail_type": "type"}}
  ],
  "minimum_confidence_threshold": 0.0 to 1.0,
  "reasoning": "why these elements are needed"
}}
Respond ONLY with the JSON object."""
def generate_ideal_schema(query, client, graph_schema):
    hub_entities = [h["entity"] for h in graph_schema["hub_entities"][:20]]
    prompt = IDEAL_SCHEMA_PROMPT.format(
        query=query,
        hub_entities=", ".join(hub_entities)
    )
    for attempt in range(3):
        try:
            r = client.models.generate_content(
                model=GENERATION_MODEL, contents=prompt,
                config={"temperature": 0.1}
            )
            raw = r.text.strip().replace("```json","").replace("```","").strip()
            return safe_parse_json(raw)
        except Exception as e:
            if "503" in str(e):
                wait = 30 * (attempt + 1)
                print(f"  503 ideal schema, waiting {wait}s (attempt {attempt+1}/3)")
                time.sleep(wait)
                continue
            print(f"  Ideal schema error: {e}")
            return None
    return None
def compute_schema_match(ideal_schema, scored_edges, retrieved_records,
                          entity_embeddings, client, alpha=0.5, beta=0.5):
    """
    Compute match score between ideal subschema and retrieved subgraph.
    Returns: overall_score, structural_score, semantic_score, breakdown
    """
    if ideal_schema is None:
        return 0.5, 0.5, 0.5, {}  # neutral score if generation failed

    # --- Structural score ---
    # Fraction of required relation types present in retrieved subgraph
    required_relations = set(ideal_schema.get("required_relations", []))
    retrieved_relations = set(e["relation"] for e in scored_edges)

    if required_relations:
        relations_found   = required_relations & retrieved_relations
        structural_score  = len(relations_found) / len(required_relations)
        missing_relations = required_relations - retrieved_relations
    else:
        structural_score  = 0.5
        relations_found   = set()
        missing_relations = set()

    # Also check ideal triplet patterns
    ideal_patterns = ideal_schema.get("ideal_triplet_patterns", [])
    pattern_scores = []
    for pattern in ideal_patterns:
        req_rel  = pattern.get("relation", "")
        # Check if any retrieved edge matches this relation type
        matched = any(e["relation"] == req_rel for e in scored_edges)
        pattern_scores.append(1.0 if matched else 0.0)

    if pattern_scores:
        pattern_score    = np.mean(pattern_scores)
        structural_score = 0.5 * structural_score + 0.5 * pattern_score

    # --- Semantic score ---
    # Embed required entity types and compare to retrieved entity embeddings
    required_entity_types = ideal_schema.get("required_entity_types", [])
    retrieved_entities    = list(set(
        [e["head"] for e in scored_edges] + [e["tail"] for e in scored_edges]
    ))[:50]  # cap for speed

    semantic_score = 0.5  # default
    if required_entity_types and retrieved_entities:
        try:
            # Embed required entity type descriptions
            req_result = client.models.embed_content(
                model=EMBEDDING_MODEL, contents=required_entity_types
            )
            req_embs = np.array([e.values for e in req_result.embeddings], dtype=np.float32)
            req_embs = req_embs / (np.linalg.norm(req_embs, axis=1, keepdims=True) + 1e-9)

            # Get embeddings for retrieved entities
            ret_embs_list = [entity_embeddings.get(e) for e in retrieved_entities
                             if entity_embeddings.get(e) is not None]
            if ret_embs_list:
                ret_embs = np.array(ret_embs_list, dtype=np.float32)
                ret_embs = ret_embs / (np.linalg.norm(ret_embs, axis=1, keepdims=True) + 1e-9)

                # For each required type, find max similarity to any retrieved entity
                sim_matrix = req_embs @ ret_embs.T  # (n_required, n_retrieved)
                max_sims   = sim_matrix.max(axis=1)  # best match per required type
                semantic_score = float(np.mean(max_sims))
        except Exception as e:
            print(f"  Semantic scoring error: {e}")

    # --- Overall score ---
    overall_score = alpha * structural_score + beta * semantic_score

    breakdown = {
        "structural_score":  round(structural_score, 4),
        "semantic_score":    round(semantic_score,   4),
        "overall_score":     round(overall_score,    4),
        "required_relations":    list(required_relations),
        "relations_found":       list(relations_found),
        "missing_relations":     list(missing_relations),
        "required_entity_types": required_entity_types,
        "pattern_match_rate":    round(np.mean(pattern_scores), 4) if pattern_scores else None
    }

    return overall_score, structural_score, semantic_score, breakdown


# Test pre-generation guardrail
test_query  = "What causes hydraulic leaks in landing gear?"
print(f"Query: {test_query}")
ideal_schema = generate_ideal_schema(test_query, client, graph_schema)
print(f"\nIdeal schema generated:")
print(json.dumps(ideal_schema, indent=2))
````

````python
POST_GUARDRAIL_PROMPT = """You are a factual grounding verifier for an aviation maintenance QA system.

QUESTION: {query}

RETRIEVED SOURCE RECORDS:
{context}

GENERATED ANSWER:
{answer}

GRAPH RETRIEVAL TRACE (edges used to retrieve records):
{trace_summary}

Evaluate the answer on three dimensions:

1. GROUNDING: Is every factual claim in the answer supported by the source records?
   - Check each claim against the records
   - Identify any claims NOT found in the records (potential hallucinations)

2. ATTRIBUTION: Can each key claim be traced to a specific record ID?
   - List which record supports which claim

3. COMPLETENESS: Does the answer address the question given what the records contain?
   - If records are insufficient, does the answer correctly say so?

Respond ONLY with JSON:
{{
  "grounding_score": 0.0 to 1.0,
  "attribution_score": 0.0 to 1.0,
  "completeness_score": 0.0 to 1.0,
  "overall_post_score": 0.0 to 1.0,
  "hallucinated_claims": ["list of claims not supported by records"],
  "well_grounded_claims": ["list of claims well supported"],
  "attribution_map": {{"claim summary": "record_id"}},
  "verdict": "PASS" or "FAIL" or "PARTIAL",
  "reasoning": "brief explanation"
}}"""

def post_generation_guardrail(query, answer, retrieved_records,
                               scored_edges, client):
    context = format_context(retrieved_records)
    trace_lines = []
    for e in scored_edges[:10]:
        trace_lines.append(
            f"{e['head']} | {e['relation']} | {e['tail']} "
            f"(score={e['relevance_score']:.3f}, "
            f"records={e['record_ids'][:3]})"
        )
    trace_summary = "\n".join(trace_lines)
    prompt = POST_GUARDRAIL_PROMPT.format(
        query=query, context=context[:3000],
        answer=answer, trace_summary=trace_summary
    )
    for attempt in range(3):
        try:
            r = client.models.generate_content(
                model=GENERATION_MODEL, contents=prompt,
                config={"temperature": 0.0}
            )
            raw = r.text.strip().replace("```json","").replace("```","").strip()
            return safe_parse_json(raw)
        except Exception as e:
            if "503" in str(e):
                wait = 30 * (attempt + 1)
                print(f"  503 post-guardrail, waiting {wait}s (attempt {attempt+1}/3)")
                time.sleep(wait)
                continue
            print(f"  Post-guardrail error: {e}")
            break
    return {"verdict": "ERROR", "overall_post_score": 0.5,
            "grounding_score": 0.5, "attribution_score": 0.5,
            "completeness_score": 0.5, "hallucinated_claims": [],
            "well_grounded_claims": [], "attribution_map": {}}


print("Post-generation guardrail function defined.")
````

````python
CONFIDENCE_THRESHOLD = 0.45  # below this: abstain or flag

GENERATION_PROMPT = """You are an aviation maintenance expert assistant.
Answer the question based ONLY on the provided maintenance records.
If the records do not contain sufficient information, say so explicitly.
Be specific and reference which records support your answer.

QUESTION: {query}

RETRIEVED MAINTENANCE RECORDS:
{context}

ANSWER:"""
def guarded_graphrag_query(query, client, G, entity_matrix_norm, entity_list,
                            entity_to_idx, entity_embeddings, relation_embeddings,
                            record_lookup, graph_schema,
                            top_k_seeds=10, hops=2, top_n_edges=30,
                            max_records=15, confidence_threshold=CONFIDENCE_THRESHOLD):
    """
    Full GraphRAG pipeline with both guardrails.
    Returns: answer, full_trace (including guardrail scores)
    """
    trace = {"query": query, "steps": {}, "guardrails": {}}

    # Step 1: Domain filter
    domain_check = check_domain(query, client)
    trace["steps"]["domain_filter"] = domain_check
    if domain_check["domain"] == "OUT_OF_DOMAIN":
        trace["steps"]["abstained"]  = True
        trace["steps"]["abstain_reason"] = "out_of_domain"
        return (f"Out of domain: {domain_check['reason']}", trace)

    # Step 2: Embed + seed nodes
    query_emb  = embed_query(query, client)
    seed_nodes = find_seed_nodes(query_emb, entity_matrix_norm, entity_list, G,
                                  top_k=top_k_seeds)
    trace["steps"]["seed_nodes"] = seed_nodes

    # Step 3: Expand + score
    sub_edges, visited = expand_subgraph(seed_nodes, G, hops=hops)
    scored_edges = score_edges_semantic(sub_edges, query_emb, entity_embeddings,
                                         relation_embeddings, top_n=top_n_edges)
    trace["steps"]["subgraph"] = {"nodes": len(visited), "edges": len(sub_edges)}
    trace["steps"]["scored_edges"] = [{"head": e["head"], "relation": e["relation"],
        "tail": e["tail"], "score": e["relevance_score"]} for e in scored_edges[:10]]

    # Step 4: Retrieve records
    retrieved = retrieve_records(scored_edges, record_lookup, max_records=max_records)
    trace["steps"]["retrieved_records"] = [{"record_id": r["record_id"],
        "source_file": r["source_file"], "source_row": r["source_row"],
        "relevance_score": r["relevance_score"]} for r in retrieved]

    # ============================================================
    # PRE-GENERATION GUARDRAIL
    # ============================================================
    ideal_schema = generate_ideal_schema(query, client, graph_schema)
    overall_score, structural_score, semantic_score, schema_breakdown = \
        compute_schema_match(ideal_schema, scored_edges, retrieved,
                              entity_embeddings, client)

    trace["guardrails"]["pre_generation"] = {
        "ideal_schema":       ideal_schema,
        "schema_match":       schema_breakdown,
        "overall_score":      overall_score,
        "structural_score":   structural_score,
        "semantic_score":     semantic_score,
        "threshold":          confidence_threshold,
        "passed":             overall_score >= confidence_threshold
    }

    # Abstain if schema match below threshold
    if overall_score < confidence_threshold:
        searched_nodes = [s["node"] for s in seed_nodes[:5]]
        missing_rels   = schema_breakdown.get("missing_relations", [])
        abstain_msg = (
            f"The knowledge graph does not contain sufficient information to answer "
            f"this query confidently (schema match score: {overall_score:.2f}, "
            f"threshold: {confidence_threshold}). "
            f"Searched around: {searched_nodes}. "
            f"Missing graph patterns: {missing_rels}."
        )
        trace["steps"]["abstained"]    = True
        trace["steps"]["abstain_reason"] = "low_schema_match"
        return abstain_msg, trace

    # Step 5: Generate answer
    if not retrieved:
        trace["steps"]["abstained"]    = True
        trace["steps"]["abstain_reason"] = "no_records_retrieved"
        return "No relevant records found.", trace

    context = format_context(retrieved)
    prompt  = GENERATION_PROMPT.format(query=query, context=context)
    r       = client.models.generate_content(model=GENERATION_MODEL, contents=prompt,
                                              config={"temperature": 0.2})
    answer  = r.text.strip()

    # ============================================================
    # POST-GENERATION GUARDRAIL
    # ============================================================
    post_result = post_generation_guardrail(
        query, answer, retrieved, scored_edges, client
    )
    trace["guardrails"]["post_generation"] = post_result

    # If post-guardrail fails: flag but still return answer with warning
    if post_result.get("verdict") == "FAIL":
        hallucinated = post_result.get("hallucinated_claims", [])
        answer = (
            f"{answer}\n\n"
            f"[GUARDRAIL WARNING: {len(hallucinated)} claim(s) could not be "
            f"verified against source records: {hallucinated[:2]}]"
        )

    trace["steps"]["abstained"] = False
    return answer, trace
````

````python
# --- All pipeline functions from notebook 4 ---

def embed_query(query, client, model=EMBEDDING_MODEL):
    result = client.models.embed_content(model=model, contents=[query])
    emb = np.array(result.embeddings[0].values, dtype=np.float32)
    return emb / (np.linalg.norm(emb) + 1e-9)

def find_seed_nodes(query_emb, entity_matrix_norm, entity_list, G,
                    top_k=10, diversity_lambda=0.5):
    similarities = entity_matrix_norm @ query_emb
    candidates = [(entity_list[i], float(similarities[i]))
                  for i in range(len(entity_list)) if entity_list[i] in G.nodes]
    candidates.sort(key=lambda x: -x[1])
    candidates = candidates[:200]
    selected, selected_embs = [], []
    while len(selected) < top_k and candidates:
        best_score, best_idx = -999, 0
        for i, (entity, sim_to_query) in enumerate(candidates):
            if not selected_embs:
                mmr_score = sim_to_query
            else:
                entity_emb = entity_matrix_norm[entity_to_idx[entity]]
                max_sim = max(float(np.dot(entity_emb, s)) for s in selected_embs)
                mmr_score = diversity_lambda * sim_to_query - (1 - diversity_lambda) * max_sim
            if mmr_score > best_score:
                best_score, best_idx = mmr_score, i
        entity, sim = candidates.pop(best_idx)
        selected.append({"node": entity, "similarity": sim, "mmr_score": best_score})
        selected_embs.append(entity_matrix_norm[entity_to_idx[entity]])
    return selected

def expand_subgraph(seed_nodes, G, hops=2):
    visited  = set(s["node"] for s in seed_nodes)
    frontier = set(visited)
    for _ in range(hops):
        new_frontier = set()
        for node in frontier:
            if node not in G: continue
            for nb in list(G.successors(node)) + list(G.predecessors(node)):
                if nb not in visited:
                    new_frontier.add(nb)
                    visited.add(nb)
        frontier = new_frontier
    subgraph_edges = []
    for u, v, data in G.edges(data=True):
        if u in visited and v in visited:
            subgraph_edges.append({"head": u, "tail": v,
                "relation": data["relation"], "record_ids": data["record_ids"],
                "support_count": data["support_count"]})
    return subgraph_edges, visited

def score_edges_semantic(subgraph_edges, query_emb, entity_embeddings,
                          relation_embeddings, top_n=30):
    scored = []
    for edge in subgraph_edges:
        head_emb = entity_embeddings.get(edge["head"])
        tail_emb = entity_embeddings.get(edge["tail"])
        rel_emb  = relation_embeddings.get(edge["relation"])
        if head_emb is None or tail_emb is None or rel_emb is None: continue
        h = np.array(head_emb, dtype=np.float32); h /= (np.linalg.norm(h) + 1e-9)
        t = np.array(tail_emb, dtype=np.float32); t /= (np.linalg.norm(t) + 1e-9)
        r = np.array(rel_emb,  dtype=np.float32); r /= (np.linalg.norm(r)  + 1e-9)
        score = 0.4 * float(np.dot(query_emb, h)) + \
                0.4 * float(np.dot(query_emb, t)) + \
                0.2 * float(np.dot(query_emb, r))
        scored.append({**edge, "relevance_score": score})
    scored.sort(key=lambda x: -x["relevance_score"])
    return scored[:top_n]

def retrieve_records(scored_edges, record_lookup, max_records=15):
    record_scores   = defaultdict(float)
    record_edge_map = defaultdict(list)
    for edge in scored_edges:
        for rid in edge["record_ids"]:
            record_scores[rid] += edge["relevance_score"]
            record_edge_map[rid].append({"head": edge["head"], "relation": edge["relation"],
                                          "tail": edge["tail"], "score": edge["relevance_score"]})
    sorted_recs = sorted(record_scores.items(), key=lambda x: -x[1])
    retrieved = []
    for rid, _ in sorted_recs[:max_records]:
        if rid in record_lookup:
            rec = record_lookup[rid]
            retrieved.append({"record_id": rid, "source_file": rec["source_file"],
                "source_row": rec.get("source_row", "unknown"),
                "text": rec["text_original"], "relevance_score": record_scores[rid],
                "supporting_edges": record_edge_map[rid][:3]})
    return retrieved

def check_domain(query, client):
    prompt = """You are a domain classifier for an aviation maintenance knowledge system.
Classify the query as IN_DOMAIN or OUT_OF_DOMAIN.
Also classify as LIKELY or UNLIKELY answerable if in-domain.
Query: "{query}"
Respond ONLY with JSON: {{"domain": "IN_DOMAIN" or "OUT_OF_DOMAIN", "answerable": "LIKELY" or "UNLIKELY", "reason": "one sentence"}}""".format(query=query)
    try:
        r = client.models.generate_content(model=GENERATION_MODEL, contents=prompt,
                                            config={"temperature": 0.0})
        raw = r.text.strip().replace("```json","").replace("```","").strip()
        return json.loads(raw)
    except:
        return {"domain": "IN_DOMAIN", "answerable": "LIKELY", "reason": "filter failed"}

def format_context(retrieved_records):
    lines = []
    for i, rec in enumerate(retrieved_records, 1):
        lines.append(f"[Record {i} | {rec['record_id']} | {rec['source_file']}]")
        lines.append(rec['text'])
        lines.append("")
    return "\n".join(lines)

print("Pipeline functions loaded.")
````

````python
def compute_metrics(results_by_qtype, threshold=0.70, system_name=""):
    metrics = {}
    all_results = []
    for qtype, results in results_by_qtype.items():
        for r in results:
            r["_qtype"] = qtype
            all_results.append(r)

    n_total = len(all_results)
    if n_total == 0:
        return {}

    # Domain accuracy
    metrics["domain_acc"] = np.mean([r.get("domain_correct", False) for r in all_results])

    # Abstention on unanswerable
    unansw = [r for r in all_results if r["_qtype"] == "in_domain_unanswerable"]
    if unansw:
        if "pre_guardrail_score" in unansw[0]:
            # Handle None scores safely
            abstained = []
            for r in unansw:
                score = r.get("pre_guardrail_score")
                below = (score is not None and score < threshold)
                abstained.append(below or r.get("abstained", False))
        else:
            abstained = [r.get("abstention_correct", False) for r in unansw]
        metrics["abstention_acc_unanswerable"] = np.mean(abstained)

    # False abstention on answerable
    answerable = [r for r in all_results
                  if r["_qtype"] in ["in_domain_answerable", "multi_hop"]]
    if answerable:
        if "pre_guardrail_score" in (answerable[0] if answerable else {}):
            false_abs = []
            for r in answerable:
                score = r.get("pre_guardrail_score")
                false_abs.append(score is not None and score < threshold)
        else:
            false_abs = [r.get("abstained", False) for r in answerable]
        metrics["false_abstention_rate"] = np.mean(false_abs)

    # OOD rejection
    ood = [r for r in all_results if r["_qtype"] == "out_of_domain"]
    if ood:
        metrics["ood_rejection"] = np.mean(
            [r.get("domain_detected", "") == "OUT_OF_DOMAIN" for r in ood])

    # Grounding score
    grounding = [r.get("post_grounding_score") for r in all_results
                 if r.get("post_grounding_score") is not None]
    if grounding:
        metrics["grounding_score"] = np.mean(grounding)

    # Hallucination rate
    halluc = [len(r.get("hallucinated_claims", [])) for r in all_results]
    metrics["hallucination_rate"] = np.mean(halluc)

    # Interception rate
    if any("pre_guardrail_score" in r for r in all_results):
        intercepted = []
        for r in all_results:
            score = r.get("pre_guardrail_score")
            intercepted.append(score is not None and score < threshold)
        metrics["interception_rate"] = np.mean(intercepted)

    # Latency
    latencies = [r.get("latency", 0) for r in all_results if r.get("latency")]
    if latencies:
        metrics["avg_latency"] = np.mean(latencies)

    # Records retrieved
    metrics["avg_records"] = np.mean([r.get("records_retrieved", 0) for r in all_results])

    return metrics


# Build comparison table
print("=== FULL SYSTEM COMPARISON ===")
print()

systems_to_compare = {
    "Vanilla LLM":        all_baseline_results["vanilla_llm"],
    "Naive RAG":          all_baseline_results["naive_rag"],
    "KEO-style KG-RAG":   all_baseline_results["keo_style"],
    "Self-RAG style":     all_baseline_results["selfrag_style"],
    "Semantic KG-RAG":    semantic_baseline_results,
    "Full System":        guarded_results,
}

all_metrics = {}
for name, results in systems_to_compare.items():
    m = compute_metrics(results, threshold=0.70, system_name=name)
    all_metrics[name] = m

# Print table
metrics_to_show = [
    ("domain_acc",                  "Domain Acc."),
    ("abstention_acc_unanswerable",  "Abstention Acc. (unansw.)"),
    ("false_abstention_rate",        "False Abstention Rate"),
    ("ood_rejection",                "OOD Rejection"),
    ("grounding_score",              "Grounding Score"),
    ("hallucination_rate",           "Halluc. Rate"),
    ("interception_rate",            "Interception Rate"),
    ("avg_latency",                  "Avg Latency (s)"),
    ("avg_records",                  "Avg Records Retrieved"),
]

# Header
sys_names = list(systems_to_compare.keys())
header = f"{'Metric':<30}" + "".join(f"{n[:15]:>17}" for n in sys_names)
print(header)
print("-" * (30 + 17 * len(sys_names)))

for metric_key, metric_label in metrics_to_show:
    row = f"{metric_label:<30}"
    for name in sys_names:
        val = all_metrics[name].get(metric_key)
        if val is None:
            row += f"{'N/A':>17}"
        elif metric_key == "avg_latency":
            row += f"{val:>16.2f}s"
        elif metric_key in ["avg_records", "hallucination_rate"]:
            row += f"{val:>17.2f}"
        else:
            row += f"{val:>16.1%}"
    print(row)

# Save metrics
with open("./output/comparison_metrics.json", "w") as f:
    json.dump(all_metrics, f, indent=2)
print("\nMetrics saved to ./output/comparison_metrics.json")
````

````python
# Recompute abstention for baseline systems with stricter criteria
# True abstention = system explicitly says it cannot answer
# NOT abstention = system gives an answer even if it hedges

STRICT_ABSTAIN_PHRASES = [
    "i cannot answer",
    "i don't have enough information",
    "this question cannot be answered",
    "no relevant records found",
    "outside the scope",
    "i am unable to answer",
    "cannot be determined from"
]

SOFT_HEDGE_PHRASES = [  # these are NOT abstentions - system still answered
    "do not contain",
    "not enough information",
    "insufficient",
    "limited information"
]

def is_strict_abstention(answer):
    """Only count as abstention if system truly refused to answer."""
    answer_lower = answer.lower()
    return any(phrase in answer_lower for phrase in STRICT_ABSTAIN_PHRASES)

# Recompute for all baseline systems
print("=== STRICT ABSTENTION RECOMPUTE ===")
for sys_name, qtypes in all_baseline_results.items():
    unansw = qtypes.get("in_domain_unanswerable", [])
    if not unansw:
        continue
    strict = sum(1 for r in unansw if is_strict_abstention(r.get("answer", "")))
    soft   = sum(1 for r in unansw if not is_strict_abstention(r.get("answer", ""))
                 and any(p in r.get("answer","").lower() for p in SOFT_HEDGE_PHRASES))
    answered = len(unansw) - strict - soft
    print(f"\n{sys_name} (n={len(unansw)}):")
    print(f"  Strict abstain (refused):  {strict}/{len(unansw)} = {strict/len(unansw):.1%}")
    print(f"  Soft hedge (answered+hedge): {soft}/{len(unansw)} = {soft/len(unansw):.1%}")
    print(f"  Answered without hedge:    {answered}/{len(unansw)} = {answered/len(unansw):.1%}")
````
